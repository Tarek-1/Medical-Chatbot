```python
from langchain.document_loaders import DirectoryLoader, PyPDFLoader

def load_pdf_files(data):
    loader = DirectoryLoader(data, glob="*.pdf", loader_cls=PyPDFLoader)
    document = loader.load()
    return document
```

```python
extracted_data = load_pdf_files("data")
```

```python
extracted_data[0:10]
```

```python
from langchain.schema import Document
from typing import List

def filter_to_minimal_docs(docs: List[Document]) -> List[Document]:
    minimal_docs = []
    for doc in docs:
        src = doc.metadata.get("source")
        minimal_docs.append(Document(page_content=doc.page_content, metadata = {"source": src}))
    return minimal_docs
```

```python
minimal_docs = filter_to_minimal_docs(extracted_data)
```

```python
minimal_docs[0:10]
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

def text_splitter(minimal_docs):    
    splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap = 20,)
    texts_chunks = splitter.split_documents(minimal_docs)
    return texts_chunks
```

```python
texts_chunks = text_splitter(minimal_docs)
```

```python
texts_chunks[0:10]
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings


def download_embeddings():

    model_name = "sentence-transformers/all-MiniLM-L6-v2"
    embeddings = HuggingFaceEmbeddings(model_name = model_name)
    return embeddings
```

```python
embedding = download_embeddings()
```

```python
# Load the index
from langchain_community.vectorstores import Pinecone

vectorstore = Pinecone.from_existing_index(
    embedding = embedding,
    index_name = index_name)
```

```python
retriever = vectorstore.as_retriever(search_type="similarity", search_kwargs={"k":3})
```

```python
retrieved_docs = retriever.invoke("What causes acne?")
retrieved_docs
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

chatModel = ChatGoogleGenerativeAI(model="models/gemini-flash-latest")
```

```python
from langchain.chains import create_retrieval_chain
from langchain.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate


system_prompt = (
    "You are a Medical assistant for question-answering tasks. "
    "Use the following pieces of retrieved context to answer "
    "the question. If you don't know the answer, say that you "
    "don't know. Use three sentences maximum and keep the "
    "answer concise.\n\n"
    "{context}"
)

prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", "{input}")
])

doc_chain = create_stuff_documents_chain(chatModel, prompt)
retrieval_chain = create_retrieval_chain(retriever, doc_chain)
```

```python
response = retrieval_chain.invoke({"input": "What causes acne?"})
print(response["answer"])
print(response["context"])
```
